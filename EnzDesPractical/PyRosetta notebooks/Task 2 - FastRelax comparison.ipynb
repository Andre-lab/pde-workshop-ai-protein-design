```python
relax_repeats_widget = widgets.IntSlider(
    value=1,
    min=1,
    max=5,
    step=1,
    description="FastRelax runs:",
    continuous_update=False
)

relax_bb_widget = widgets.Checkbox(
    value=True,
    description="Allow backbone movement"
)

relax_chi_widget = widgets.Checkbox(
    value=True,
    description="Allow side-chain movement"
)

relax_jump_widget = widgets.Checkbox(
    value=False,
    description="Allow rigid-body jumps"
)

display(
    relax_repeats_widget,
    relax_bb_widget,
    relax_chi_widget,
    relax_jump_widget
)
```

```python
def run_fastrelax(
    pose,
    repeats=1,
    allow_bb=True,
    allow_chi=True,
    allow_jump=False
):

    relax = rosetta.protocols.relax.FastRelax()

    relax.set_scorefxn(
        scorefxn
    )

    # Define movable degrees of freedom
    movemap = pyrosetta.MoveMap()

    movemap.set_bb(
        allow_bb
    )

    movemap.set_chi(
        allow_chi
    )

    movemap.set_jump(
        allow_jump
    )

    relax.set_movemap(
        movemap
    )

    # Apply FastRelax
    for _ in range(repeats):

        relax.apply(pose)

    return pose
```

```python
def get_per_residue_energies(
    pose,
    scorefxn
):

    # Make sure energies are up to date
    scorefxn(pose)

    energies = pose.energies()

    weights = scorefxn.weights()

    residue_numbers = []
    residue_names = []
    residue_energies = []

    for i in range(
        1,
        pose.total_residue() + 1
    ):

        # Weighted total energy for residue i
        energy = (
            energies
            .residue_total_energies(i)
            .dot(weights)
        )

        residue_numbers.append(i)

        residue_names.append(
            pose.residue(i).name3()
        )

        residue_energies.append(
            float(energy)
        )

    return pd.DataFrame({
        "Pose_residue": residue_numbers,
        "Residue": residue_names,
        "Energy_REU": residue_energies
    })
```

```python
def load_structure(path):

    if not os.path.exists(path):

        raise FileNotFoundError(
            path
        )

    pose = pyrosetta.pose_from_file(
        path
    )

    return pose
```

```python
# ============================================================
# FASTRELAX BATCH PROCESSING FOR TWO STRUCTURES
# ============================================================

run_button = widgets.Button(
    description="Run FastRelax",
    button_style="success",
    icon="play"
)

run_output = widgets.Output()

# Results stored globally so the download buttons can use them
relaxed_poses = {}
starting_scores = {}
final_scores = {}
per_residue_results = {}
latest_summary = pd.DataFrame()


def run_analysis(_):

    global relaxed_poses
    global starting_scores
    global final_scores
    global per_residue_results
    global latest_summary

    with run_output:

        clear_output()

        # --------------------------------------------------------
        # 1. Check that both structures have been uploaded
        # --------------------------------------------------------

        if structure1_path is None:
            print("ERROR: Structure 1 has not been uploaded.")
            return

        if structure2_path is None:
            print("ERROR: Structure 2 has not been uploaded.")
            return

        if not os.path.exists(structure1_path):
            print(
                f"ERROR: Structure 1 cannot be found:\n"
                f"{structure1_path}"
            )
            return

        if not os.path.exists(structure2_path):
            print(
                f"ERROR: Structure 2 cannot be found:\n"
                f"{structure2_path}"
            )
            return

        # --------------------------------------------------------
        # 2. Build structure list
        # --------------------------------------------------------

        structures = [
            {
                "index": 1,
                "path": structure1_path,
                "name": structure1_original_name
            },
            {
                "index": 2,
                "path": structure2_path,
                "name": structure2_original_name
            }
        ]

        # Clear previous results
        relaxed_poses = {}
        starting_scores = {}
        final_scores = {}
        per_residue_results = {}

        summary_rows = []

        print("=" * 70)
        print("FASTRELAX ANALYSIS")
        print("=" * 70)

        print(
            f"FastRelax repeats: "
            f"{relax_repeats_widget.value}"
        )

        print(
            f"Backbone movement: "
            f"{relax_bb_widget.value}"
        )

        print(
            f"Side-chain movement: "
            f"{relax_chi_widget.value}"
        )

        print(
            f"Jump movement: "
            f"{relax_jump_widget.value}"
        )

        print()

        # ========================================================
        # 3. Process structure 1 and structure 2
        # ========================================================

        for structure in structures:

            index = structure["index"]
            path = structure["path"]
            original_name = structure["name"]

            print()
            print("=" * 70)
            print(
                f"STRUCTURE {index}/2: "
                f"{original_name}"
            )
            print("=" * 70)

            try:

                # ------------------------------------------------
                # Load structure
                # ------------------------------------------------

                print("Loading structure...")

                pose = pyrosetta.pose_from_file(
                    path
                )

                print(
                    f"Loaded {pose.total_residue()} "
                    f"residues."
                )

                # ------------------------------------------------
                # Starting score
                # ------------------------------------------------

                start_score = float(
                    scorefxn(pose)
                )

                print(
                    f"Starting score: "
                    f"{start_score:.3f}"
                )

                # ------------------------------------------------
                # Run FastRelax
                # ------------------------------------------------

                print(
                    "Running FastRelax..."
                )

                pose = run_fastrelax(
                    pose,
                    repeats=(
                        relax_repeats_widget.value
                    ),
                    allow_bb=(
                        relax_bb_widget.value
                    ),
                    allow_chi=(
                        relax_chi_widget.value
                    ),
                    allow_jump=(
                        relax_jump_widget.value
                    )
                )

                # ------------------------------------------------
                # Final score
                # ------------------------------------------------

                final_score = float(
                    scorefxn(pose)
                )

                score_change = (
                    final_score
                    - start_score
                )

                print(
                    f"Relaxed score: "
                    f"{final_score:.3f}"
                )

                print(
                    f"Score change: "
                    f"{score_change:.3f}"
                )

                # ------------------------------------------------
                # Save in memory
                # ------------------------------------------------

                relaxed_poses[index] = pose
                starting_scores[index] = (
                    start_score
                )
                final_scores[index] = (
                    final_score
                )

                # ------------------------------------------------
                # Per-residue energies
                # ------------------------------------------------

                per_residue_df = (
                    get_per_residue_energies(
                        pose,
                        scorefxn
                    )
                )

                per_residue_results[index] = (
                    per_residue_df
                )

                # ------------------------------------------------
                # Save relaxed PDB
                # ------------------------------------------------

                base_name = os.path.splitext(
                    original_name
                )[0]

                output_name = (
                    f"{base_name}.relaxed.pdb"
                )

                output_path = os.path.join(
                    "/content",
                    output_name
                )

                pose.dump_pdb(
                    output_path
                )

                print(
                    f"Relaxed structure saved: "
                    f"{output_name}"
                )

                # ------------------------------------------------
                # Summary row
                # ------------------------------------------------

                summary_rows.append({

                    "Structure":
                        original_name,

                    "Format":
                        os.path.splitext(
                            original_name
                        )[1].lower(),

                    "N_residues":
                        pose.total_residue(),

                    "Starting_score_REU":
                        start_score,

                    "Relaxed_score_REU":
                        final_score,

                    "Score_change_REU":
                        score_change,

                    "Relaxed_PDB":
                        output_name,

                    "Status":
                        "Success"
                })

                print(
                    "Structure complete."
                )

            except Exception as e:

                error_message = (
                    f"{type(e).__name__}: {e}"
                )

                print()
                print(
                    f"FAILED: {error_message}"
                )

                summary_rows.append({

                    "Structure":
                        original_name,

                    "Format":
                        os.path.splitext(
                            original_name
                        )[1].lower(),

                    "N_residues":
                        np.nan,

                    "Starting_score_REU":
                        np.nan,

                    "Relaxed_score_REU":
                        np.nan,

                    "Score_change_REU":
                        np.nan,

                    "Relaxed_PDB":
                        "",

                    "Status":
                        f"Failed: {error_message}"
                })

        # ========================================================
        # 4. Summary table
        # ========================================================

        latest_summary = pd.DataFrame(
            summary_rows
        )

        print()
        print("=" * 70)
        print("FASTRELAX SUMMARY")
        print("=" * 70)

        display(
            latest_summary
        )

        # Save summary CSV
        latest_summary.to_csv(
            "/content/fastrelax_summary.csv",
            index=False
        )

        print(
            "\nSaved: /content/fastrelax_summary.csv"
        )

        # ========================================================
        # 5. Plot per-residue energies
        # ========================================================

        if len(per_residue_results) > 0:

            plt.figure(
                figsize=(12, 6)
            )

            for index, df in (
                per_residue_results.items()
            ):

                original_name = (
                    structure1_original_name
                    if index == 1
                    else structure2_original_name
                )

                plt.plot(
                    df["Pose_residue"],
                    df["Energy_REU"],
                    label=original_name,
                    linewidth=1.5
                )

            plt.xlabel(
                "Pose residue number"
            )

            plt.ylabel(
                "Per-residue energy (REU)"
            )

            plt.title(
                "Per-residue energy after FastRelax"
            )

            plt.legend()

            plt.grid(
                alpha=0.3
            )

            plt.tight_layout()

            plt.show()

        # ========================================================
        # 6. Save combined per-residue energy CSV
        # ========================================================

        if len(per_residue_results) > 0:

            combined_rows = []

            for index, df in (
                per_residue_results.items()
            ):

                original_name = (
                    structure1_original_name
                    if index == 1
                    else structure2_original_name
                )

                temp = df.copy()

                temp.insert(
                    0,
                    "Structure",
                    original_name
                )

                combined_rows.append(
                    temp
                )

            combined_energy_df = pd.concat(
                combined_rows,
                ignore_index=True
            )

            combined_energy_df.to_csv(
                "/content/"
                "per_residue_energies.csv",
                index=False
            )

            print(
                "Saved: "
                "/content/per_residue_energies.csv"
            )

        print()
        print("=" * 70)
        print("DONE")
        print("=" * 70)


run_button.on_click(
    run_analysis
)

display(
    run_button,
    run_output
)
```

```python
def save_relaxed_structures(_):

    with save_output:

        clear_output()

        if not relaxed_poses:
            print(
                "No relaxed structures available. "
                "Run FastRelax first."
            )
            return

        structure_names = {
            1: structure1_original_name,
            2: structure2_original_name
        }

        for index, pose in relaxed_poses.items():

            original_name = structure_names[index]

            base_name = os.path.splitext(
                original_name
            )[0]

            output_name = (
                f"{base_name}.relaxed.pdb"
            )

            output_path = os.path.join(
                "/content",
                output_name
            )

            pose.dump_pdb(
                output_path
            )

            print(
                f"Saved: {output_name}"
            )
```

```python
download_button = widgets.Button(
    description="Download relaxed structures",
    button_style="info",
    icon="download"
)

download_output = widgets.Output()


def download_relaxed(_):

    with download_output:

        clear_output()

        for index, pose in relaxed_poses.items():

            original_name = (
                structure1_original_name
                if index == 1
                else structure2_original_name
            )

            base_name = os.path.splitext(
                original_name
            )[0]

            output_name = (
                f"{base_name}.relaxed.pdb"
            )

            output_path = os.path.join(
                "/content",
                output_name
            )

            pose.dump_pdb(
                output_path
            )

            print(
                f"Downloading {output_name}"
            )

            files.download(
                output_path
            )


download_button.on_click(
    download_relaxed
)

display(
    download_button,
    download_output
)
```
